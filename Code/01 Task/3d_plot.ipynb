```python
from import_data import data_import
import pandas as pd
```

```python
features = ["F1", "F2", "F3", "F4", "F5", "F6", "F7", "F8"]
files = ["SNR_noiseless"]

data, sample_labels = data_import(files, features)
full_data = pd.DataFrame(data, columns=features)

# Make a 3D scatter of F3, F5, F8, with one legend entry per sample label.
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

sample_label_series = pd.Series(sample_labels)
for label in sample_label_series.unique():
    mask = sample_label_series == label
    ax.scatter(
        full_data.loc[mask, 'F3'],
        -full_data.loc[mask, 'F5'],
        full_data.loc[mask, 'F8'],
        label=label,
    )

ax.set_xlabel('F3')
ax.set_ylabel('F5')
ax.set_zlabel('F8')
# put the legend outside the plot
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
```
