```python
from import_data import data_import
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from classifier import Classifier
import matplotlib.pyplot as plt
import pandas as pd

# Data used
files = ["SNR_20db"] 

# Import data and create a DataFrame
features = ["F1", "F2", "F3", "F4", "F5", "F6", "F7", "F8"]
data, labels = data_import(files, features)
full_data = pd.DataFrame(data, columns=features)

# add labels to the dataframe
full_data['Label'] = labels

# plot all features vs labels in a 4x2 grid
plt.figure(figsize=(16, 16))
for i, feature in enumerate(features):
    plt.subplot(4, 2, i + 1)
    plt.scatter(full_data[feature], full_data['Label'], cmap='viridis', alpha=0.5)
    plt.xlabel(feature)
    plt.ylabel('Label')
    plt.title(f'{feature} vs Label')

plt.tight_layout()
```
